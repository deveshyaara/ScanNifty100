```python
import pandas as pd
from pathlib import Path

SOURCE_DIR = Path("../data/source")

files = [
    "analysis.xlsx",
    "balancesheet.xlsx",
    "cashflow.xlsx",
    "companies.xlsx",
    "documents.xlsx",
    "profitandloss.xlsx",
    "prosandcons.xlsx"
]

def inspect_file(file):
    path = SOURCE_DIR / file
    df = pd.read_excel(path)

    print(f"\n📄 {file}")
    print("="*50)
    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())
    print("\nSample:")
    display(df.head(3))
    print("\nDtypes:")
    print(df.dtypes)
    print("\nNull %:")
    print((df.isnull().mean() * 100).round(2))

for f in files:
    inspect_file(f)
```
